# Annotation statistics: agreement score 

## agreement score before resolving mismatches

```python
yash_csv_path = "../ground_truth/annotation_agreement/set2&3_02_23_yash.csv"
chen_csv_path = "../ground_truth/annotation_agreement/set2&3_02_23_chen.csv"

df_yash = pd.read_csv(yash_csv_path)
df_chen = pd.read_csv(chen_csv_path)

df_yash = df_yash[df_yash['annotator'].notna() & (df_yash['annotator'].astype(str).str.strip() != "")]
df_yash = df_yash.rename(columns={
    "verification": "yash_classification",
    "explanation_verification": "yash_explanation"
})
df_yash = df_yash[['data_broker', 'prompting_strategy', 'status', 'pattern_name', 'yash_classification', 'yash_explanation']]

df_chen = df_chen[df_chen['annotator'].notna() & (df_chen['annotator'].astype(str).str.strip() != "")]
df_chen = df_chen.rename(columns={
    "verification": "chen_classification",
    "explanation_verification": "chen_explanation"
})
df_chen = df_chen[['data_broker', 'prompting_strategy', 'status', 'pattern_name', 'chen_classification', 'chen_explanation']]

df_merged = pd.merge(df_yash, df_chen, on=['data_broker', 'prompting_strategy', 'status', 'pattern_name'], how='inner')
df_merged_sorted = df_merged.sort_values(
    by="data_broker",
    ascending=True
)
kappa = cohen_kappa_score(
    df_merged["yash_classification"],
    df_merged["chen_classification"]
)

print(f"Cohen's Kappa: {kappa:.3f}")
```

## agreement score after resolving the mismatches 

```python
df_set1 = pd.read_csv('../ground_truth/annotation_agreement/set1_12_16_result.csv')
df_set2 =pd.read_csv('../ground_truth/annotation_agreement/set2&3_02_23_result.csv')
df_set2 = df_set2[['Y_classification_2', 'C_classification_2']]
df_set1 = df_set1[['Y_classification_2', 'C_classification_2']]

df_all = pd.concat([df_set1, df_set2], ignore_index=True)
kappa = cohen_kappa_score(
    df_all["Y_classification_2"],
    df_all["C_classification_2"]
)

print(f"Cohen's Kappa: {kappa:.3f}")
```
